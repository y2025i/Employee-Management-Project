```python
from sqlalchemy import create_engine, Column, String, Integer 
```

```python
from sqlalchemy.ext.declarative import declarative_base
```

```python
from sqlalchemy.orm import sessionmaker
```

```python
engine= create_engine("sqlite:///employees.db",connect_args={'check_same_thread': False}) #create engine
```

```python
Base = declarative_base() #createbase of the tables
```

```python
class Employee(Base): #create table. id is gingto be our primary key
    __tablename__='employees'
    id=Column(Integer,primarykey=True)
    name = Column(String)
    department = Column(String)
    position = Column(String)
    salary = Column(String) #or integer?
```
